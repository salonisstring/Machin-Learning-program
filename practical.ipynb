```python
x=df[["Daily Time Spent on Site", "Daily Internet Usage"]]#Features
y=df["Clicked on Ad"]#Target
print("Features:")
print(x.head())
print("\nTarget:")
print(y.head())
```

```python
#split data
from sklearn.model_selection import train_test_split
x_train, x_test,y_train,y_test = train_test_split(x,y,test_size=0.20,random_state=42,stratify=y)
print("Training dat: ",x_train.shape)
print("Teating data:",x_test.shape)
```

```python
#Init scaler
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)#Scale training
x_test_scaled = scaler.transform(x_test)#scale test data
```

```python
x_test_scaled
```

```python
x_test_sc=pd.DataFrame(x_test_scaled)
x_test_sc.head()
x_test_sc.describe()
```

```python
#import model
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()

#Train model
model.fit(x_train_scaled,y_train)
```

```python
#Predicts labels
y_pred = model.predict(x_test_scaled)

#Predict probabilities
y_probability = model.predict_proba(x_test_scaled)[:,1]

print("Predicted classes:")
print(y_probability[:10])
```
